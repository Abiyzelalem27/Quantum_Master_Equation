## Lindblad master equation


In classical physics, a *master equation* commonly
describes the evolution of a probability distribution. For our quantum
system, the **quantum master equation** evolves the entire density matrix
$\rho_S$, so it describes both populations (diagonal entries) and
coherences (off-diagonal entries).

An equation for the populations alone is often called a **rate equation**.

### Lindblad master equation

Starting from an initial system state $\rho_S(0)=\rho_0$, the density
matrix evolves according to

$$
\boxed{
\frac{d\rho_S}{dt}
=
-i[H,\rho_S]
+
\sum_\alpha
\left(
\Gamma_\alpha\rho_S\Gamma_\alpha^\dagger
-\frac12\Gamma_\alpha^\dagger\Gamma_\alpha\rho_S
-\frac12\rho_S\Gamma_\alpha^\dagger\Gamma_\alpha
\right).
}
$$

The two contributions have different meanings:

- **Hamiltonian term:** $-i[H,\rho_S]$ describes coherent (unitary)
  evolution.
- **Lindblad terms:** the sum describes environmental effects, such as
  spontaneous emission or dephasing.

The operators $\Gamma_\alpha$ are called **jump operators**. Each one
represents a possible dissipative process. For example, decay from the
excited state $|e\rangle$ to the ground state $|g\rangle$ is described by

$$
\Gamma=\sqrt{\gamma}\,|g\rangle\langle e|,
$$

where $\gamma$ is the decay rate.

We can also write the equation compactly using a **Liouvillian
superoperator** $\mathcal L$:

$$
\frac{d\rho_S}{dt}=\mathcal L[\rho_S].
$$

A superoperator takes an operator as its input and returns another
operator: $\rho_S\mapsto\mathcal L[\rho_S]$.

**Convention:** The equation above uses $\hbar=1$. With $\hbar$ shown
explicitly, the Hamiltonian term is $-\frac{i}{\hbar}[H,\rho_S]$.

## spontaneous emission of a two-level atom

We consider an atom with ground state $|g\rangle$ and excited state
$|e\rangle$. The atom begins in the excited state and can decay to
$|g\rangle$ by emitting a photon.

We order the basis as $(|g\rangle,|e\rangle)$ and use $\hbar=1$
throughout the calculation. The Hamiltonian, jump operator, and
initial state are

$$
H=\omega|e\rangle\langle e|,
\qquad
\Gamma_{e\to g}=\sqrt{\gamma}\,|g\rangle\langle e|,
\qquad
\rho(0)=|e\rangle\langle e|.
$$

Here $\omega$ is the transition angular frequency and $\gamma>0$ is
the spontaneous-emission rate. Equivalently, before setting
$\hbar=1$, the Hamiltonian is
$H=\hbar\omega|e\rangle\langle e|$.

### Step 1: Start from the Lindblad equation

For one jump operator $\Gamma$, the master equation is

$$
\dot\rho
=
-i[H,\rho]
+
\Gamma\rho\Gamma^\dagger
-
\frac12
\left(
\Gamma^\dagger\Gamma\rho+
\rho\Gamma^\dagger\Gamma
\right).
$$

The commutator is $[H,\rho]=H\rho-\rho H$. The first term describes
the atom's coherent evolution. The other terms describe its decay.

### Step 2: Substitute the Hamiltonian

Since $H=\omega|e\rangle\langle e|$,

$$
-i[H,\rho]
=
-i\omega
\left(
|e\rangle\langle e|\rho
-
\rho|e\rangle\langle e|
\right).
$$

Keep the two products in this order: matrix multiplication generally
does not commute.

### Step 3: Calculate the jump term

The adjoint of the jump operator is

$$
\Gamma^\dagger
=
\sqrt{\gamma}\,|e\rangle\langle g|.
$$

Therefore,

$$
\begin{aligned}
\Gamma\rho\Gamma^\dagger
&=
\gamma
|g\rangle\langle e|
\rho
|e\rangle\langle g|\\
&=
\gamma
\underbrace{\langle e|\rho|e\rangle}_{\rho_{ee}}
|g\rangle\langle g|\\
&=
\gamma\rho_{ee}|g\rangle\langle g|.
\end{aligned}
$$

The number $\rho_{ee}=\langle e|\rho|e\rangle$ is the excited-state
population. Thus this term adds population to $|g\rangle$ at a rate
proportional to the population still in $|e\rangle$.

### Step 4: Calculate $\Gamma^\dagger\Gamma$

Using $\langle g|g\rangle=1$,

$$
\begin{aligned}
\Gamma^\dagger\Gamma
&=
\gamma
|e\rangle\langle g|g\rangle\langle e|\\
&=
\gamma|e\rangle\langle e|.
\end{aligned}
$$

Consequently,

$$
\Gamma^\dagger\Gamma\rho
=
\gamma|e\rangle\langle e|\rho,
\qquad
\rho\Gamma^\dagger\Gamma
=
\gamma\rho|e\rangle\langle e|.
$$

Substituting Steps 2–4 gives the operator equation reached in the
handwritten calculation:

$$
\boxed{
\begin{aligned}
\dot\rho
={}&-i\omega
\left(
|e\rangle\langle e|\rho
-
\rho|e\rangle\langle e|
\right)\\
&+\gamma\rho_{ee}|g\rangle\langle g|\\
&-\frac{\gamma}{2}
\left(
|e\rangle\langle e|\rho
+
\rho|e\rangle\langle e|
\right).
\end{aligned}
}
$$

We now extract each matrix element. Define

$$
\rho_{ij}=\langle i|\rho|j\rangle,
\qquad i,j\in\{g,e\}.
$$

### Step 5: Excited-state population

Multiply the operator equation on the left by $\langle e|$ and on
the right by $|e\rangle$:

$$
\dot\rho_{ee}=\langle e|\dot\rho|e\rangle.
$$

The Hamiltonian part is zero because its two products both give
$\rho_{ee}$:

$$
-i\omega(\rho_{ee}-\rho_{ee})=0.
$$

The jump contribution is zero because $\langle e|g\rangle=0$.
Each subtraction product gives $\rho_{ee}$. Hence

$$
\begin{aligned}
\dot\rho_{ee}
&=
0+0-\frac{\gamma}{2}
(\rho_{ee}+\rho_{ee})\\
&=\boxed{-\gamma\rho_{ee}}.
\end{aligned}
$$

Separate variables and integrate:

$$
\frac{d\rho_{ee}}{\rho_{ee}}
=
-\gamma\,dt,
\qquad
\ln\rho_{ee}(t)-\ln\rho_{ee}(0)=-\gamma t.
$$

Therefore,

$$
\rho_{ee}(t)=\rho_{ee}(0)e^{-\gamma t}.
$$

Because $\rho(0)=|e\rangle\langle e|$, we have
$\rho_{ee}(0)=1$, so

$$
\boxed{\rho_{ee}(t)=e^{-\gamma t}.}
$$

The excited-state probability falls exponentially from $1$ toward $0$.

### Step 6: Ground-state population

This time multiply by $\langle g|$ and $|g\rangle$:

$$
\dot\rho_{gg}=\langle g|\dot\rho|g\rangle.
$$

The Hamiltonian and subtraction terms are zero because they contain
$|e\rangle\langle e|$ next to a ground-state bra or ket. The jump
term gives $\gamma\rho_{ee}$. Thus

$$
\boxed{\dot\rho_{gg}=\gamma\rho_{ee}
=\gamma e^{-\gamma t}.}
$$

The atom begins excited, so $\rho_{gg}(0)=0$. Integrating from $0$
to $t$,

$$
\begin{aligned}
\rho_{gg}(t)-\rho_{gg}(0)
&=\int_0^t\gamma e^{-\gamma s}\,ds\\
&=\left[-e^{-\gamma s}\right]_{s=0}^{s=t}\\
&=1-e^{-\gamma t}.
\end{aligned}
$$

Therefore,

$$
\boxed{\rho_{gg}(t)=1-e^{-\gamma t}.}
$$

The ground-state probability grows by exactly the amount lost from
the excited state.

### Step 7: Off-diagonal coherence

For the $ge$ element, take
$\dot\rho_{ge}=\langle g|\dot\rho|e\rangle$.

The Hamiltonian term contributes $+i\omega\rho_{ge}$; the jump term
contributes zero; and the subtraction terms together contribute
$-\gamma\rho_{ge}/2$. Hence

$$
\boxed{
\dot\rho_{ge}
=
\left(i\omega-\frac{\gamma}{2}\right)\rho_{ge}.
}
$$

Its general solution is

$$
\rho_{ge}(t)
=
\rho_{ge}(0)e^{(i\omega-\gamma/2)t}.
$$

Similarly,

$$
\rho_{eg}(t)
=
\rho_{eg}(0)e^{(-i\omega-\gamma/2)t}.
$$

For our initial state $|e\rangle\langle e|$, both coherences are
initially zero:

$$
\rho_{ge}(0)=\rho_{eg}(0)=0.
$$

Multiplying zero by the exponential still gives zero, so

$$
\boxed{\rho_{ge}(t)=\rho_{eg}(t)=0.}
$$

The equations show how *existing* coherences would decay. They do
not create coherence from this initially excited state.

### Step 8: Assemble the density matrix

The basis order is $(|g\rangle,|e\rangle)$, so

$$
\rho(t)=
\begin{pmatrix}
\rho_{gg}(t)&\rho_{ge}(t)\\
\rho_{eg}(t)&\rho_{ee}(t)
\end{pmatrix}.
$$

Substituting our four results,

$$
\boxed{
\rho(t)
=
\begin{pmatrix}
1-e^{-\gamma t}&0\\
0&e^{-\gamma t}
\end{pmatrix}
=
\left(1-e^{-\gamma t}\right)|g\rangle\langle g|
+
e^{-\gamma t}|e\rangle\langle e|.
}
$$

### Step 9: Check the answer

At $t=0$,

$$
\rho(0)=
\begin{pmatrix}0&0\\0&1\end{pmatrix}
=
|e\rangle\langle e|,
$$

which matches the stated initial condition.

The trace remains one:

$$
\operatorname{Tr}\rho(t)
=
\rho_{gg}(t)+\rho_{ee}(t)
=
1-e^{-\gamma t}+e^{-\gamma t}
=
1.
$$

The two eigenvalues are $1-e^{-\gamma t}$ and
$e^{-\gamma t}$. Both lie between $0$ and $1$ for $t\geq0$, so
$\rho(t)$ is positive semidefinite.

Finally,

$$
\lim_{t\to\infty}\rho(t)
=
|g\rangle\langle g|.
$$

Thus the atom starts excited, gradually loses its excitation to the
environment, and approaches the ground state.

**Connection to the earlier decay channel:** The probability that
decay has occurred by time $t$ is

$$
\boxed{p(t)=1-e^{-\gamma t}.}
$$

For the initially excited state, $\rho_{ge}(0)=0$, so
$\rho_{ge}(t)=0$. Hermiticity gives $\rho_{eg}(t)=0$ as well.

Putting the populations and coherences together,

$$
\begin{aligned}
\rho(t)
&=
\left(1-e^{-\gamma t}\right)|g\rangle\langle g|
+
e^{-\gamma t}|e\rangle\langle e|\\[2mm]
&=
\boxed{
\begin{pmatrix}
1-e^{-\gamma t}&0\\
0&e^{-\gamma t}
\end{pmatrix}_{(|g\rangle,|e\rangle)}.
}
\end{aligned}
$$

**Physical interpretation**

- Population transfers from the excited state $|e\rangle$ to the
  ground state $|g\rangle$:
  $\rho_{ee}(t)=e^{-\gamma t}$ and
  $\rho_{gg}(t)=1-e^{-\gamma t}$.
- If coherence is initially present, its magnitude decays as
  $e^{-\gamma t/2}$. For the initially excited atom, coherence
  starts at zero and stays zero.
- The trace is preserved:
  $\operatorname{Tr}\rho(t)=\rho_{gg}(t)+\rho_{ee}(t)=1$.
- The density matrix remains positive: its eigenvalues
  $1-e^{-\gamma t}$ and $e^{-\gamma t}$ are nonnegative for $t\geq0$.
